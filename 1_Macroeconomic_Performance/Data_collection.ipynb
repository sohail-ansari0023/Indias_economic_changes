2. **How fast did India's real GDP grow?**
- Annual real GDP gowth rate

- Compound growth of real GDP
- Compare the growth trajectory rather than only the starting and ending values

````python
# Collecting the GDP info from World bank database with the help of wbgapi.

# These are the indicators code.
indi_codes: dict = {
    'GDP growth (annual %)': 'NY.GDP.MKTP.KD.ZG',
    'GDP (current US$)': 'NY.GDP.MKTP.CD',
    'GDP (constant 2015 US$)': 'NY.GDP.MKTP.KD',
    'GDP, PPP (current international $)': 'NY.GDP.MKTP.PP.CD',
    'GDP per capita (current US$)': 'NY.GDP.PCAP.CD',
    'Inflation, GDP deflator (annual %)': 'NY.GDP.DEFL.KD.ZG'
}

# let's extract data
gdp_df: pd.DataFrame = wb.data.DataFrame(
    list(indi_codes.values()),
    'IND',
    time=range(2004, 2025)
)

gdp_df
````

````python
# let's copy the data set quickly so we don't have to run the extraction again.
df: pd.DataFrame = gdp_df.copy()
````

### **Start preparing and preprocessing them**
**we'll follow the Industry steps.**

````python
# Transpose the dataframe so we can make year as a column which
# contains each year and the other indices as columns too.

df: pd.DataFrame = df.T
df.head()
````

**A minor but necessary thing**

````python
df = df.reset_index() # Shifted the index into a columns
df = df.rename_axis(None, axis=0).rename_axis(None, axis=1) # removing the axis name of the column headers.
````

**Structural Standardization**

(Normalizing the columns names) since we have only 7 of them We'll use a standard list of names.

````python
# Structural standardization

columns: list = [
    'Year',
    'Inflation(annual %)',
    'GDP_Current(US$ Trillion)',
    'GDP_Constant(2015 US$ Trillion)',
    'GDP_Growth(annual%)',
    'GDP_PPP(Current US$ Trillion)',
    'GDP_Per_Capita(Current US$)'
]

df.columns = columns

df.head()
````

**Deduplication**

I don't think that there is any..but for being cautious.

````python
# Checking for duplicate values.

df.duplicated().sum()
````

None like I said.

**Missing values**

Don't think that could be also true. Stilllll

````python
# Checking for missing values
df.isna().sum()
````

None... clear!

**Datatype Casting**

we'll let's **normalize** each columns first.

````python
# Replace the 'YR' string from the Year column and convert them into an integer.
df['Year'] = df['Year'].str.replace('YR', '')
df['Year'] = df['Year'].astype('Int64')

# Inflation(annual%) let's round it to 2 decimal places
df['Inflation(annual %)'] = df['Inflation(annual %)'].round(2)

# GDP_Current, GDP_Constant, and GDP_PPP let's round them into 2 decimal places too,
# Since they are in trillions we gotta handle them accordingly.abs

columns: list = ['GDP_Current(US$ Trillion)', 'GDP_Constant(2015 US$ Trillion)', 'GDP_PPP(Current US$ Trillion)']

for col in columns:
    df[col] = (df[col] / 1_000_000_000_000).round(2)

# Round the remaining GDP_Growth(annual%) and GDP_Per_Capita(Current US$)
df[['GDP_Growth(annual%)','GDP_Per_Capita(Current US$)']] = df[['GDP_Growth(annual%)','GDP_Per_Capita(Current US$)']].round(2)
````

each of the datatype is correct....

**Exporting**

While exporting I'm encountring the following errors:
> No type extension with name arrow.py_extension_type found

it's because simply PyArrow doesn't support **Pandas** special types, such as:
- Pandas special nullable integer(`Int64`) or floating types, or `strings`

convert them back to standard float64 before exporting, which PyArrow handles perfectly

Another error:
> A type extension with name pandas.period already defined
The error A type extension with name pandas.period already defined occurs because the PyArrow engine is trying to register a specific time-related data format into its internal memory, but finds that the format is already listed there.

#### **Let's collect**

- **Aggregate of World GDP from 2004-2024**
- **GDP Current for two major economy _China_ and _USA_.**
- **GDP, PPP (Current international $) for each country.**

````python
# The Indicators and their codes.
indicators: dict = {
    'GDP (current US$)': 'NY.GDP.MKTP.CD',
    'GDP, PPP (current international $)': 'NY.GDP.MKTP.PP.CD',
}

df_3 = wb.data.DataFrame(
    list(indicators.values()),
    ['WLD', 'CHN', 'USA'],
    time=range(2004, 2025)
)

df_3
````

### **Transform and Clean the data**

````python
df3 = df_3.copy() # making the copy of the dataframe
````

````python
df3 = df3.T # Transposing the Dataframe
df3 = df3.reset_index() # reset the index label
````

#### **Structural Standardization**

Since our dataframe contains multiindex columns, let's standardize them.

````python
# columns name list
columns_list: list = [
    'Year',
    'CHN_GDP_Current(US$ Trillion)',
    'CHN_GDP_PPP(current International US$ Trillion)',
    'USA_GDP_Current(US$ Trillion)',
    'USA_GDP_PPP(current International US$ Trillion)',
    'WLD_GDP_Current(US$ Trillion)',
    'WLD_GDP_PPP(current International US$ Trillion)',
]

# Assigning the proper column names to each columns.
df3.columns = columns_list

df3.head(2)
````

#### **let's clean each columns accordingly**

````python
# Strip the "YR" String from each year value and assign it an 'int' datatype.
df3['Year'] = df3['Year'].str.replace('YR', '').astype(int)

# Now for the remaining columns we must convert them in trillions till two
# decimal places.
remaining_columns: list = df3.columns[1:] # just exclude the year column.

# To convert them into trillion we must divide them with 1_000_000_000_000
for col in remaining_columns:
    df3[col] = (df3[col] / 1_000_000_000_000).round(2)

df3.head()
````

````python
df3.info()
````

### **All good! now let's merge them with our previous df_ver02** 

````python
# Ingesting the version 02 dataset, without error handling which saves time
# for this particular project.
df_ver_02 = pd.read_parquet('Cleaned_datasets/Production/production_dataset_ver02.parquet')
````

````python
df_ver03 = pd.merge(df_ver_02, df3, on='Year')

df_ver03.head(3)
````

- Production (parquet)

````python
production_file_path: str = 'Cleaned_datasets/Production/production_dataset_ver03.parquet'
stakeholders_file_path: str = 'Cleaned_datasets/Stakeholders/stakeholders_dataset_ver03.csv'
try:
    df_ver03.to_parquet(production_file_path, index=False)
    print(f'[SUCCESS] exported [df_ver03] to {production_file_path}')
    df_ver03.to_csv(stakeholders_file_path, index=False)
    print(f'[SUCCESS] exported [df_ver03] to {stakeholders_file_path}')
except Exception as e:
    print(f'[CRITICAL ERROR] couldnot export: {e}')
````
