```python
# let's retrieve some data. GDP (current US$)
df = wb.data.DataFrame(
    'NY.GDP.MKTP.KD', # code for GDP(current US$)
    'IND', # country ISO3 code
    time=range(2010, 2024)
)

type(df)
```

```python
ori_df = df.copy()
```

```python
ori_df = df.T # (flips rows and columns)
# ori_df
```

```python
ori_df = ori_df.reset_index()
ori_df
```

```python
ori_df = ori_df.rename(columns={'index': 'Year', 'IND': 'GDP ($ Trillion)'})

ori_df
```

economy is one of the most notoriously annoying quirks in Pandas. The word economy that you see floating above the index numbers isn't actually a column or a row—it is an axis name.   When you transposed and reset the index earlier, Pandas held onto the original label ("economy") and attached it as a metadata name for your new column headers. Because it isn't a physical column or row of data, .drop() cannot see it, which is why your code is throwing an error.

```python
ori_df = ori_df.rename_axis(None, axis=1).rename_axis(None, axis=0)

ori_df
```

```python
# let's clean the columsn first
ori_df['Year'] = ori_df['Year'].str.strip().str.replace('YR', '')

# Now we know the GDP value is in 'Trillion' let's round that up into 2 decimal places.
ori_df['GDP ($ Trillion)'] = ori_df['GDP ($ Trillion)'] / 1_000_000_000_000

ori_df['GDP ($ Trillion)'] = ori_df['GDP ($ Trillion)'].round(2)
ori_df
```

```python
# visualization

import matplotlib.pyplot as plt

gdp_current = ori_df.copy()

fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(
    gdp_current['Year'],
    gdp_current['GDP ($ Trillion)'],
    marker='o'
)

ax.set_title('GDP (current $) 2010 - 2023')
ax.set_xlabel('Year')
ax.set_ylabel('GDP ($ Trillion)')

ax.grid()

plt.show()
```
