This entire project is based on a concept called **Financial Stability**.

**Simple Definition**
> **Financial Stability** means a condition in which the financial system of a country—it's banks, markets, payment systems, and financial institutions—can perform its core functions smoothly, even when it faces _shocks or stress_.

Let's break down every word:
| **Term** | **Meaning** |
|----------|-------------|
| _Financial System_ | The network of banks, stock markets, insurance companies, payment systems, and regulations that move money through an economy |
| _Core Functions_ | Accepting deposits, giving loans, processing payments, allocating capital (directing money to productive uses) |
| _Shocks_ | Sudden unexpected events— a bank fails, a pandemic hits, a currency crashes, interest rates spike |
| _Stress_ | Prolonged pressure on the system—rising bad loans, falling asset prices, loss of confidence |
